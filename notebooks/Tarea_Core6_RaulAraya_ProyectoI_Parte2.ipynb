5. **Filtrado de Datos.**

    <u>Requerimiento</u>
    + Filtra el DataFrame para mostrar solo las filas donde las ventas (Ventas) sean mayores a 50.
    + Filtra el DataFrame para mostrar solo las filas donde el precio (Precio) sea menor a 0.5.
    + Utilizando el método query(), filtra el DataFrame para mostrar las filas donde el producto sea Manzanas y las ventas sean mayores a 30.

    <u>Solución</u>
    + En el siguiente código se presenta el procesamiento de los requerimientos.
    + Mediante filtro "(df['Total Amount']>50)", se presentan los datos bajo la condición de que las ventas Totales sean superior a 50, las ventas que cumplen esta condición son 783.
    + Los precios del DataFrame se mueven entre 25 y 500, por lo tanto se modifico el filtro a menor a 50. Mediante filtro "(df['Price per Unit']<50)", se presentan los datos bajo la condición de que los precios sean menores a 50, las ventas que cumplen esta condición son 393.
    + El DataFrame no pose el datos manzanas, por lo tanto se seleccione el producto "Electronics". Mediante el query "df.query('`Product Category` == "Electronics" & `Total Amount` > 30')", se presentan los datos bajo la condición de que las ventas totales sean superior a 30 y que el producto sea "Electronics", las ventas que cumplen esta condición son 783.


```python
#Filtra el DataFrame para mostrar solo las filas donde las ventas (Total Amount) sean mayores a 50.
print("Se presenta solo las ventas mayores a 50 en Total Amount")
print("="*80)
calc1=df[(df['Total Amount']>50)]
print(calc1)
print("-"*80)

#Filtra el DataFrame para mostrar solo las filas donde el precio (Price per Unit) sea menor a 50.
print("Se presenta solo las ventas menores a 50 en Price per Unit")
print("="*80)
calc2=df[(df['Price per Unit']<50)]
print(calc2)
print("-"*80)

#Utilizando el método query(), filtra el DataFrame para mostrar las filas donde el producto sea Electronics y las ventas sean mayores a 30.
print("Se presenta las ventas totales que son mayores a 1 millon y solo para el genero 'Electronics'")
print("="*80)
calc3 = df.query('`Product Category` == "Electronics" & `Total Amount` > 30')
print(calc3)
print("-"*80)
```

5. **Slicing de Datos.**

    <u>Requerimiento</u>
    + Selecciona y muestra solo las columnas Name y Global_Sales del DataFrame.
    + Utilizando loc[], selecciona y muestra las filas de la 5 a la 10 (inclusive) y las columnas Name y Genre.
    + Utilizando iloc[], selecciona y muestra las primeras 5 filas y las primeras 3 columnas del DataFrame.

    <u>Solución</u>
    + En el siguiente código se presenta el procesamiento de los requerimientos.
    + El DataFrame no posee la columana Name, por lo tanto se cambio por Age. Mediante la selelcción df[['Age', 'Total Amount']], mostramos solo los datos de las columnas 'Age' y 'Total Amount'.
    + El DataFrame no posee la columana Name, por lo tanto se cambio por Age. Mediante la selección df.loc[4:11, ['Age', 'Gender']], mostramos solo los datos de la quinta fila a la décima y solo para las columnas 'Age' y 'Gender',.
    + Mediante la selección df.iloc[0:5, 0:3], mostramos solo lsa primeras 5 filas de las 3 primeras columnas.

```python
# Selecciona y muestra solo las columnas Age y Total Amount del DataFrame.
print("Se presenta solo las columnas que llevan el nombre 'Age' y 'Total Amount'")
print("="*80)
calc4 = df[['Age', 'Total Amount']]
print(calc4)
print("-"*80)

# Utilizando loc[], selecciona y muestra las filas de la 5 a la 10 (inclusive) y las columnas Age y Gender.
print("Se presenta solo de la fila 5 a la 10, que llevan el nombre 'Age' y 'Gender'")
print("="*80)
calc5 = df.loc[4:11, ['Age', 'Gender']]
print(calc5)
print("-"*80)


# Utilizando iloc[], selecciona y muestra las primeras 5 filas y las primeras 3 columnas del DataFrame.
print("Se presenta solo lsa primeras 5 filas de las 3 primeras columnas")
print("="*80)
calc6 = df.iloc[0:5, 0:3]
print(calc6)
print("-"*80)
```
