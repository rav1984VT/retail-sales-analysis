### Paso 3 Exploración de Datos Avanzada

#### 3.1 Analisis de correlación y mapa de calor.

- La variable objetivo del estudio es 'Total Amount'.
- Esta variable tiene una correlación positiva con la variable 'Price por Unit' y en menor medida con la variable 'Quantity'. Con la variable 'Age' existe una correlación negativa y muy baja.

```python
# Paso 1: calcular la correlación
cols_num = ["Transaction ID", "Age", "Quantity", "Price per Unit", "Total Amount"]
corr = df[cols_num].corr()

print("Matriz de correlación:")
display(corr.round(2))

# Paso 2: dibujarla con heatmap
fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,          # muestra el número dentro de cada celda
    fmt=".2f",           # formato del número
    cmap="RdYlGn",       # rojo=negativo, verde=positivo
    center=0,
    vmin=-1, vmax=1,
    square=True,
    linewidths=0.5,
    annot_kws={"size": 12},
    ax=ax
)

ax.set_title("Correlación ventas de reatil", fontsize=14, fontweight="bold", pad=16)
plt.tight_layout()
plt.show()
```

#### 3.2 Subplots.

Gráficos de dispersión

- A continuación, se muestra dos gráficos de dispersión que muestran la relación entre las variables 'Price per Unit' y 'Quantity' con la variable 'Total Amount'.
- Ambos gráficos muestran una relación positiva.
- Para precios bajos la dispersión de valors es muy baja, a medida que el precio aumenta la dispersión con los ingresos aumenta.
- La indicado en el punto anterior afecta directamente a la dispersión entre la variable cantidad e ingresos. Donde para ingresos bajos la dispersión es baja, pero para ingresos altos la dispersión es alta.

```python
# Crear gráficos de dispersión para analizar la relación entre 'Price per Unit' y 'Total Amount', y entre 'Quantity' y 'Total Amount'
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Primer gráfico de dispersión
axes[0].scatter(df['Total Amount'], df['Price per Unit'], color='blue', alpha=0.7,label='Price per Unit')
axes[0].set_title('Relación Price per Unit y Total Amount')
axes[0].set_xlabel('Total Amount')
axes[0].set_ylabel('Price per Unit')
axes[0].legend()
axes[0].annotate('Relación positiva', xy=(190, 50), xytext=(750, 55),
            arrowprops=dict(facecolor='black', shrink=0.05))
axes[0].grid(True, which='major', linestyle='--', linewidth=0.5, color='gray')

# Segundo gráfico de dispersión
axes[1].scatter(df['Total Amount'], df['Quantity'], color='green', alpha=0.7,label='Quantity')
axes[1].set_title('Relación Quantity y Total Amount')
axes[1].set_xlabel('Total Amount')
axes[1].set_ylabel('Quantity')
axes[1].legend(loc='lower right')
axes[1].arrow(1000, 2, 650, -.9, head_width=.1, head_length=40, fc='green', ec='green')
axes[1].grid(True, which='major', linestyle='--', linewidth=0.5, color='gray')


plt.tight_layout()
plt.show()
```

Graficos de Histogramas.

- La mayor cantidad de ingresos por ventas se concentran entre en el rango de 0-250.
- Los tres grupos de productos 'Electronics', 'Beauty' y 'Clothing' presentan un comportamiento similar de ingresos por ventas.
- Se observa un sesgo hacia la derecha.

```python
# Crear gráficos de histograma para analizar la distribución de 'Total Amount', con y sin KDE, y separado por 'Product Category'
# utilizando el parámetro hue

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Distribución Ventas Totales — Retail", fontsize=14, fontweight="bold")

# Izquierda: histograma simple
sns.histplot(data=df, x="Total Amount", bins=10,
             color="#6C3CE1", ax=axes[0])
axes[0].set_title("Solo histograma")
axes[0].set_xlabel("Total Amonut (unidades monetarias)")
axes[0].annotate('Sesgo hacia la derecha', xy=(250, 200), xytext=(600, 400),
            arrowprops=dict(facecolor='black', shrink=0.05))
axes[0].grid(True, which='major', linestyle='--', linewidth=0.5, color='gray')

# Derecha: histograma + KDE + separado por especie
sns.histplot(data=df, x="Total Amount", hue="Product Category",
             kde=True, bins=10, alpha=0.5, ax=axes[1])
axes[1].set_title("Con KDE y por categoria (hue)")
axes[1].set_xlabel("Total Amonut (unidades monetarias)")
axes[1].annotate('KDE', xy=(200, 75), xytext=(500, 100),
            arrowprops=dict(facecolor='black', shrink=0.05))
axes[1].grid(True, which='major', linestyle='--', linewidth=0.5, color='gray')

plt.tight_layout()
plt.show()
```
