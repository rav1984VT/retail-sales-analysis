### Paso 3 Exploración de Datos Avanzada

#### 3.1 Analisis de correlación y mapa de calor.

```python
# Paso 1: calcular la correlación
cols_num = ["Transaction ID", "Age", "Quantity", "Price per Unit", "Total Amount"]
corr = df[cols_num].corr()

print("Matriz de correlación:")
display(corr.round(2))

# Paso 2: dibujarla con heatmap
fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,          # muestra el número dentro de cada celda
    fmt=".2f",           # formato del número
    cmap="RdYlGn",       # rojo=negativo, verde=positivo
    center=0,
    vmin=-1, vmax=1,
    square=True,
    linewidths=0.5,
    annot_kws={"size": 12},
    ax=ax
)

ax.set_title("Correlación ventas de reatil", fontsize=14, fontweight="bold", pad=16)
plt.tight_layout()
plt.show()
```

#### 3.2 Subplots.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Distribución Ventas Totales — Retail", fontsize=14, fontweight="bold")

# Izquierda: histograma simple
sns.histplot(data=df, x="Total Amount", bins=10,
             color="#6C3CE1", ax=axes[0])
axes[0].set_title("Solo histograma")
axes[0].set_xlabel("Total Amonut (unidades monetarias)")

# Derecha: histograma + KDE + separado por especie
sns.histplot(data=df, x="Total Amount", hue="Product Category",
             kde=True, bins=10, alpha=0.5, ax=axes[1])
axes[1].set_title("Con KDE y por categoria (hue)")
axes[1].set_xlabel("Total Amonut (unidades monetarias)")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Product Category",
    y="Quantity",
    hue="Gender",
    palette="husl",
    width=0.6,
    ax=ax
)

ax.set_title("Cantidades vendidas por Categoría y Sexo", fontsize=14, fontweight="bold", pad=16)
ax.set_xlabel("Product Category")
ax.set_ylabel("Cantidad (unidades)")
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Gráfico de dispersión
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time', ax=axs[0, 0])
axs[0, 0].set_title('Total Bill vs Tip')

# Gráfico de barras
sns.barplot(data=df, x='day', y='total_bill', hue='sex', ax=axs[0, 1])
axs[0, 1].set_title('Total Bill por Día y Sexo')

# Gráfico de cajas
sns.boxplot(data=df, x='day', y='total_bill', hue='smoker', ax=axs[1, 0])
axs[1, 0].set_title('Distribución de Total Bill por Día y Fumadores')

# Histograma
sns.histplot(data=df, x='total_bill', kde=True, ax=axs[1, 1])
axs[1, 1].set_title('Histograma de Total Bill')

# Ajustar la posición de los subplots para evitar superposición
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatter(
    data=df,
    x="Total Amount",
    y="Price per Unit",
    hue="Category",
    palette="husl",
    width=0.6,
    ax=ax
)

ax.set_title("Cantidades vendidas por Categoría y Sexo", fontsize=14, fontweight="bold", pad=16)
ax.set_xlabel("Product Category")
ax.set_ylabel("Cantidad (unidades)")
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
pair = sns.pairplot(
    df,
    hue="Product Category",
    diag_kind="kde",
    plot_kws={"alpha": 0.5, "s": 30}
)
pair.figure.suptitle("Pairplot — Medidas de Pingüinos", y=1.02, fontsize=14, fontweight="bold")
plt.show()
```
