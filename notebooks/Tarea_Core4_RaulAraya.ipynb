**3-Exploración de Datos:**
-   Calcula el total de ventas por categoría de producto.
-   Calcula el promedio de ventas diarias por categoría de producto.
-   Identifica las categorías de productos con mayores y menores ventas.

<u>Solución:</u>

En el siguiente código, se presentan los resultados obtenidos, con la finalidad de explorar el data frame cargado.

```python
#Calcula el total de ventas por categoría de producto.
total_ventas_por_categoria=df.groupby(['Product Category'])['Quantity'].agg(['sum'])
print("Total Ventas por Categoria")
print("="*30)
print(total_ventas_por_categoria)
print("-"*30)
```

```python
#Calcula el promedio de ventas diarias por categoría de producto.
ventas_dia=df.groupby(['Product Category','Date'])['Quantity'].agg(['sum'])
print("Total ventas diarias por producto")
print("="*40)
print(ventas_dia)
print("-"*40)

print("Ventas promedio diarias por producto")
print("="*40)
ventas_prom_dia=ventas_dia.groupby(['Product Category'])['sum'].agg(['mean']).round(2)
print(ventas_prom_dia)
print("-"*40)
```

```python
#Identifica las categorías de productos con mayores y menores ventas.
total_ventas=df.groupby(['Product Category'])['Quantity'].agg(['sum'])
print("Total Ventas por Categoria")
print("="*30)
print(total_ventas)
print("-"*30)

print("Prodcutor con mayor ventas")
print("="*30)
filas_maximas = total_ventas[total_ventas['sum'] == total_ventas['sum'].max()]
print(filas_maximas)
print("-"*30)

print("Prodcutor con menor ventas")
print("="*30)
filas_minimas = total_ventas[total_ventas['sum'] == total_ventas['sum'].min()]
print(filas_minimas)
print("-"*30)
```

**4-Manipulación de Datos:**
-   Filtra los datos para mostrar solo las ventas de una categoría de producto específica.
-   Realiza operaciones de suma, resta, multiplicación y división en los datos para obtener estadísticas adicionales.

<u>Solución:</u>

En el siguiente código, se presentan los resultados obtenidos, con la finalidad de realizar una manipulación del data frame cargado.

```python
#Filtra los datos para mostrar solo las ventas de una categoría de producto específica.

#Primero se filtra solo el producto Electronics
filtro_grupo = df.groupby(['Product Category']).filter(lambda x: (x['Product Category'] == 'Electronics').any())

#Luego se imprime solo el nombre del procuto y su venta.
print("Total Ventas por Categoria Específica")
print("="*30)
print(filtro_grupo[['Product Category','Quantity']])
print("-"*30)
```

```python
#Realiza operaciones de suma, resta, multiplicación y división en los datos para obtener estadísticas adicionales.

#Realizamos operación suma para obtener el total de ventas, para ello primero determinamos el total de ventas por categoria con groupby
#y luego realizamos operaciones sumas para obetener el total de ventas. Este valor se utilizará luego para obtener el porcentaje
#de ventas de cada venta asociada a cada producto.
print("Total Ventas")
print("="*30)
total_ventas=df.groupby(['Product Category'])['Quantity'].agg(['sum'])
beauty=total_ventas.iat[0,0]
Clothing=total_ventas.iat[1,0]
Electronics=total_ventas.iat[2,0]
total_ventas = beauty+Clothing+Electronics
print(f"Las ventas totales son: {total_ventas}")
print("-"*30)
#valor

#Realizamos una operación resta, para lo cual agregamos un columna que es año de nacimiento, que se obtiene como
#la resta entre:(i) Año actual y (ii) la Columna "Age"
print("Agregando columna con año de nacimiento")
print("="*30)
año_actual=2026
df["Year of bith"]=año_actual - df['Age']
print(df)
print("-"*30)

#Realizamos una operación de multiplicación, la cual tiene la finalidad de determinar cuando debe aumentar la cantindad de
#ventas del producto Beauty, para que la repartición de ingresos entre los 3 productos sean similares.
print("Multiplicación")
print("="*30)
total_amount_cat1=df.groupby(['Product Category'])['Total Amount'].agg(['sum'])
total_amount1=df['Total Amount'].sum()
total_amount_cat1["Percentage Amount"]=round(total_amount_cat1['sum']/total_amount1,3)
print(total_amount1)
print(total_amount_cat1)
print("-"*30)

factores = {'Beauty': 1.1, 'Clothing': 1, 'Electronics': 1}

df2=df
df2['Quantity'] = df.apply(lambda row: row['Quantity'] * factores.get(row['Product Category'], 1), axis=1)
df2['Total Amount'] = df2['Quantity']*df2['Price per Unit']
total_amount_cat2=df2.groupby(['Product Category'])['Total Amount'].agg(['sum'])
total_amount2=df2['Total Amount'].sum()
total_amount_cat2["Percentage Amount"]=round(total_amount_cat2['sum']/total_amount2,3)
print(total_amount2)
print(total_amount_cat2)
print("-"*30)

#Realizamos operación división, para lo cual agregamos un columna que es Porcentaje de ventas totales, que se obtiene como
#la división entre:(i) la Columna "Quantity" y (ii) las ventas totales obtenidas anteriormente.
print("Agregando columna con porcenataje de ventas totales")
print("="*30)
df["Sales percentage Total"]=round(df['Quantity']/total_ventas,5)
print(df)
print("-"*30)

```
