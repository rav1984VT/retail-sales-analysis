### **Paso 2** Creación de columnas.

-   Se crearon las siguientes dos columnas al DataFrame:
    + Porcentaje de ventas (Total Amount) para cada registro, para cada grupo de la columna 'Product Category'
    + Porcentaje de ventas (Toal Amount) para cada registro.

```python
# Calcular el porcentaje de ventas (Total Amount;TA) para cada Producto (Product Category;PC)
df['Percentage_TA_PC'] = df.groupby(['Product Category'])[['Total Amount']].transform(lambda x: x / x.sum() * 100).round(2)

# Calcular el porcentaje de ventas totales
df['Percentage_TA'] = df['Total Amount'].transform(lambda x: x / x.sum() * 100).round(2)

print(df)
```

### **Paso 3** Clasificación de datos.

-   Realizamos una clasificación de las ventas (monetario), con la condición de si las ventas son superiores a 1200 es considerada un venta 'Alta', si es menor a 900 es una venta 'Baja', de lo contrario es una venta 'Media'.

```python
# Clasificar los productos por rango de ventas (monetario)
df['Classification Total Amount'] = df['Total Amount'].apply(lambda x: 'Alta' if x > 1200 else ('Baja' if x < 900 else 'Media'))
print(df)

# Guardar el DataFrame en un archivo CSV
#df.to_csv('../data/ventas_pruebaa.csv', index=False)
```

## 2. Agrupación y Agregación

- <u>Requerimiento</u>
    + Agrupación por múltiples columnas: Realiza agrupaciones por categorías como producto y tienda, producto y mes, etc.
    + Aplicar funciones de agregación: Utiliza funciones como sum, mean, count, min, max, std, y var para obtener estadísticas descriptivas de cada grupo.
    
- <u>Solución</u>
    + Realizamos en primera instancia la agrupación del DataFrame bajo el siguiente criterio:
        + Agrupación por la columna 'Product Category'.
        + Agrupación primero por la columna 'Product Category' y luego por la columna 'Gender'.
        + Esto se realiza con la finalidad de aplicar a las siguientes funciones.
    + Realizamos la aplicación de funciones para las agrupaciones antes indicadas:
        + Suma del total de ventas y del porcentaje de ventas totales, para la primera agrupación. Nos entrega la referencia del producto relevante para la empresa o si las todas los productos tienen una relevancia importante
        + Determinación de datos estadísticos para la segunda agrupación. Nos entre el comportamiento estadístico de las ventas de cada producto según el genero de quien lo compra.



```python
# Agrupaciones

#agrupamos por 'Product Category (PC)'
grouped_1 = df.groupby('Product Category')
#Agrupamos por 'Product Category' y por 'Gender'
grouped_2 = df.groupby(['Product Category','Gender'])

#Aplicación de funciones de agregación

ventas_PC=grouped_1[['Total Amount','Percentage_TA']].agg(['sum']).round(0)
ventas_Gender_PC=grouped_2[['Total Amount']].agg(['count','sum','mean','min','max','std','var']).round(1)

print("Resumen del Total de Ventas y Porcentaje de Ventas")
print("-"*100)
print(ventas_PC)
print("")
print("="*100)
print("")
print("Resumen de variables estadísticas para el Total de Ventas por Categoria de Productos")
print("-"*100)
print(ventas_Gender_PC)
```

## 3. Análisis Personalizado con apply

- <u>Requerimiento</u>
    + Función personalizada: Aplica funciones personalizadas para realizar análisis específicos que no se pueden lograr con las funciones de agregación estándar.
    + Ejemplo de uso avanzado: Calcula la desviación de cada venta respecto a la media de su grupo..
    
- <u>Solución</u>
    + Se la función personalizada correspondiente a la desviación absoluta media, para la columna 'Price Per Unit'. Esta nos permitirá medir la dispersión (distanca prommedio) entre cada registro y la medía aritmética de los precios del DataFrame.

```python
#Funciones apply

# Calcular la desviación absoluta media
mean_Price_Per_Unit = df['Price per Unit'].mean()
df['MAD_Price_Per_Unit'] = df['Price per Unit'].apply(lambda x: np.abs(x - mean_Price_Per_Unit))
print(df)
```
